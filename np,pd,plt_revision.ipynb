```python
data = {
  'Name': ['Nischal', 'Pradip', 'Sovit', 'Piyush'],
  'Age': [19, 20, 21, 22],
  'Address': ['Dang', 'Kathmandu', 'Bhaktapur', 'Pokhara'],
  'Qualification': ['BCA', 'MCA', 'BTech', 'MTech']
}
df = pd.DataFrame(data)
df
```

```python
# Common Applications of Pandas and DataFrames:
# 1. Data Cleaning & Manipulation (handling missing data, renaming columns, converting types)
# 2. Data Exploration & Analysis (value counts, statistics, grouping, sorting)
# 3. Data Ingestion (reading from CSV, Excel, SQL databases, JSON)
# 4. Merging and Joining (combining multiple datasets together)

# Let's illustrate a few key applications using our existing DataFrame 'df':

# Application 1: Quick Descriptive Statistics
print("--- Summary Statistics ---")
print(df.describe(include='all'))

# Application 2: Accessing and Slicing columns
print("\n--- Accessing Columns ---")
print(df[['Name', 'Qualification']])

# Application 3: Filtering Data (boolean indexing)
print("\n--- Filtering: Age > 20 ---")
print(df[df['Age'] > 20])

# Application 4: Adding New Columns
print("\n--- Adding New Column ---")
df['Is_Adult'] = df['Age'] >= 18
print(df)
```
